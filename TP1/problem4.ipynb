### Part B.
Write a function safe_int(text) that turns a string into an int. It must accept spaces around
the number, negative numbers and decimals ("3.9" → 3), and return None when the text is not a
number.

```python
def safe_int(text):
    """Turn text into an int; return None when it is not a number."""
    text = text.strip()
    if text.lstrip("-").isdigit():
        return int(text)
    try:
        return int(float(text))
    except ValueError:
        return None
    
for sample in ["42", " -8 ", "3.9", "abc", ""]:
    print(repr(sample), "->", safe_int(sample))
```

### Part C. 
Ask the user for two numbers with input(), convert them with safe_int() and print their
sum, or a friendly message if one of them is not a number.
Hint: str.strip(), str.isdigit() and a try / except ValueError block are all you need

```python
numbers = []
for i in range(2):
    num = input("Enter a number: ")
    numbers.append(safe_int(num))

if None in numbers:
    print("Please enter valid numbers.")
else:
    print("Sum:", sum(numbers))
```
